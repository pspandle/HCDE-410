### 4. Synthesis & Reflection

Both my probe and observation showed how easily AI generated information can seem correct even when it isn't. In my probe, I found that using the same prompt could still produce different results and factual errors. In my observation, the participant initially thought the output was correct because the table was easy to read and the assumptions stated in the output made sense. This made me think about how the way information is presented can influence how much people trust it, even when it hasn't been verified. 

Something I found interesting from both experiences was how much catching errors depends on the user. I was able to recognize the incorrect venue classification in my probe because I knew where The Concourse Project was located, but the participant said they wouldn't have caught the error because they didn't recognize all the venue names. I also noticed a difference in how the models supported their answers. Claude relied on its own knowledge when identifying venue locations, while ChatGPT provided a source link when it corrected its response. However, having a source doesn't automatically mean an answer is accurate. People still need to be able to check whether the source actually supports what the model is saying. 

These experiences made me think more about what makes research trustworthy and reproducible. If someone can't understand how an answer was produced or verify the information behind it, it becomes harder to know whether the results are reliable. I also think verification needs to be accessible, since it is difficult to check an answer when the original information is hard to understand or compare. This is especially important when AI is used for research or tasks where people might rely on the results to make decisions.
